```python
llm = OllamaLLM(
    model="tinyllama:latest",
    temperature=0.6
)

template = PromptTemplate.from_template(
    input_variable = ['cuisine'],
    template = "I want to open a restaurant for {cuisine} food. Suggest a fency name for this",
)

new_chain = LLMChain(llm=llm, prompt=template)

template2 = PromptTemplate.from_template(
    input_variable = ['restaurant_name'],
    template = "Suggest some menu items for {restaurant_name}. Return it as a comma seperated",
)
food_items_chain = LLMChain(llm=llm, prompt=template2)
```

```python
# simplesequential chain is for the single answer in sequence in chain
from langchain_classic.chains import SimpleSequentialChain

chain = SimpleSequentialChain(
    chains=[new_chain, food_items_chain],
)
response = chain.run("Italian")
print(response)
```

```python
# sequential chain is for multiple answers in sequence in chaaian
llm = OllamaLLM(
    model="tinyllama:latest",
    temperature=0.6
)

# first chain
template3 = PromptTemplate.from_template(
    input_variable = ['cuisine'],
    template = "I want to open a restaurant for {cuisine} food. Suggest a fency name for this",
)
name_chain = LLMChain(llm=llm, prompt=template3, output_key="restaurant_name")

# second chain
template4 = PromptTemplate.from_template(
    input_variable = ['restaurant_name'],
    template = "Suggest some menu items for {restaurant_name}. Return it as a comma seperated",
)
food_items_chain = LLMChain(llm=llm, prompt=template4, output_key="menu_items")
```

```python
from  langchain_classic.chains import SequentialChain

chain = SequentialChain(
    chains = [name_chain, food_items_chain],
    input_variables = ['cuisine'],
    output_variables = ['restaurant_name', 'menu_items'],
)
chain({'cuisine':'Indian'})
```

```python
### Agents
### load tools is to load the tools(3rd party), for more info visit the lanchain agetns tools site
from langchain_classic.agents import AgentType, initialize_agent
from langchain_community.agent_toolkits.load_tools import load_tools
from langchain_ollama import OllamaLLM

llm = OllamaLLM(
    model="tinyllama:latest",
    temperature=0.6
)
## the tools's we'll give the agent access to. Note that llm=math toll uses an LLM
tools = load_tools(["wikipedia","llm-math"], llm=llm)

## Finally, Let's initialize an agent with the tools, the language model, and the type of agent we want
agent = initialize_agent(tools,llm,agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION)

agent.run("When was Elon musk born? What is his age right now in 2026?")
```

```python
import os
from dotenv import load_dotenv
from langchain_ollama import OllamaLLM
load_dotenv()

serpapi = os.getenv("SERPAPI_API_KEY")

os.environ["SERPAPI_API_KEY"] = serpapi

llm = OllamaLLM(
    model="tinyllama:latest",
    temperature=0.6
)
```

```python
from langchain_community.agent_toolkits.load_tools import load_tools
from langchain_classic.agents import AgentType, initialize_agent

## the tools's we'll give the agent access to. Note that llm=math toll uses an LLM
tools = load_tools(["serpapi","llm-math"], llm=llm)

## Finally, Let's initialize an agent with the tools, the language model, and the type of agent we want 
agent = initialize_agent(tools, llm, agent=AgentType.ZERO_SHOT_REACT_DESCRIPTION, verbose=True,
    handle_parsing_errors=(
        "Use exactly this format when calling a tool:\n"
        "Action: Search\n"
        "Action Input: your search query\n"
        "When finished, use Final Answer: your answer."
    ),
    max_iterations=3
)

response = agent.run("what is the gdp of India in 2026")
print(response)
```
